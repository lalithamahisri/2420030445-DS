```python
# drop_duplicates
import pandas as pd
df=pd.DataFrame({'ID':[1,2,3,4,5],
                'Name':['Alice','Bob','Charlie','Alice','Eve'],
                'Age':[25,30,35,25,40],
                 })
print("Original DataSet (with duplicate values):")
print(df,"\n")
df_exact=df.drop_duplicates()
print("DataSet after removing exact duplicate values:")
print(df_exact)
```

```python
# droping_rows_with_missing_vals
import pandas as pd
import numpy as np 
df=pd.DataFrame({'Age':[25,30,np.nan,40,35], 'Department': ['HR', 'Finance','Finance', np.nan, 'IT']})
print(df,"\n")
df_drop_rows=df.dropna()
print("DataSet after dropping rows with missing values:")
print(df_drop_rows)
```

```python
# fwd_bwd_fill
import pandas as pd
import numpy as np 
df=pd.DataFrame({'Age':[25,30,np.nan,40,35], 'Department': ['HR', 'Finance','Finance', np.nan, 'IT']})
print("Original DataSet (with missing values):")
print(df,"\n")
df_ffill=df.copy()
df_ffill.ffill(inplace=True)
print("DataSet after Forward Fill:")
print(df_ffill)

#backward fill
df_bfill=df.copy()  
df_bfill.bfill(inplace=True)
print("DataSet after Backward Fill:")
print(df_bfill)
```

```python
# missing_vals
import pandas as pd
import numpy as np 
df=pd.DataFrame({'Age':[25,30,np.nan,40,35], 'Department': ['HR', 'Finance','Finance', np.nan, 'IT']})
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
# case_normalization_upper_lower
import pandas as pd
df=pd.DataFrame({'Name':['Alice','BOB','charlie','DAVID']})
print("Original DataSet (with inconsistent case formats):")
print(df,"\n")
df['Name_lower']=df['Name'].str.lower()
print("DataSet after converting to lowercase:")
print(df)
df['Name_upper']=df['Name'].str.upper()
print("DataSet after converting to uppercase:")
print(df)
```

```python
# correcting_inconsistent_formats
import pandas as pd
df=pd.DataFrame({'Date':['2025-01-05','05/01/2025','Jan 5, 2025','2025.01.05']})
print("Original DataSet (with inconsistent date formats):")
print(df,"\n")
df['Date']=pd.to_datetime(df['Date'], errors='coerce').dt.strftime('%Y-%m-%d')
print("DataSet after correcting inconsistent date formats:")    
print(df)
```

```python
# drop_col_with_missing_vals
import pandas as pd
import numpy as np 
df=pd.DataFrame({'Age':[25,30,np.nan,40,35], 'Department': ['HR', 'Finance','Finance', np.nan, 'IT']})
print(df,"\n")
df_drop_col=df.dropna(axis=1)
print("DataSet after dropping columns with missing values:")
print(df_drop_col)
```

```python
# drop_dup_based_on_id
import pandas as pd
df=pd.DataFrame({'ID':[1,2,2,3,4,4,5],
                 'Name':['Alice','Bob','Bob','Charlie','David','David','David'],
                'Age':[25,30,30,35,40,40,40]})
print("Original DataSet (with duplicate values):")
print(df,"\n")
df_subset_id=df.drop_duplicates(subset=['ID'])
print("DataSet after removing duplicates based on 'ID':")
print(df_subset_id) 
df_subset_name=df.drop_duplicates(subset=['Name'])
print("DataSet after removing duplicates based on 'Name':")     
print(df_subset_name)
```
