```python
#1. Univariate
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

#load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data,columns=iris.feature_names)
df['species'] = iris.target

#1. Univariate Visualization

#histogram
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'],bins=20,kde=True)
plt.title("Histogram of Sepla Length")
plt.show()


#Boxplot
plt.figure(figsize=(6,4))
sns.boxplot(df['sepal length (cm)'])
plt.title("Boxplot of Sepla Length")
plt.show()


#Pie chart
species_coumts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_coumts,labels=iris.target_names, autopct='%1.1f%%')
plt.title("Species distibution")
plt.show()
```

```python
#2. Bivariate Visulaizations

#Bivariate visualizations
#Scatter Plot

plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'], palette='Set1')
plt.title('Sepal Length vs Sepal Width')
plt.show()

#Line chart
#Line Chart
plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title('Trend of Sepal Length')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()

#bar chart
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'], y=df['sepal length (cm)'])
plt.title("Bar chart: mean sepal length by species")
plt.xticks([0,1,2], iris.target_names)
plt.show()
```

```python
#3. Multivariate Visualizations
#heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:, :4], annot=True, cmap='coolwarm')
plt.title('Heatmap of feature correlations')
plt.show()

#bubble chart
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'],y=df['sepal width (cm)'], s=df['petal length (cm)']*20, c=df['species'], alpha=0.5)
plt.title('Bubble Chart: Sepal Length vs Sepal Width with Petal Length as Size')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()

#pair plot(relationships among all features)
sns.pairplot(df.iloc[:,:4])
plt.suptitle('Pair Plot of Iris Features', y=1.02)
plt.show()
```

```python
#4.Advanced Techniques
#violin plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'], y=df['sepal length (cm)'])
plt.title('Violin Plot: Sepal Length by Species')
plt.xticks([0,1,2], iris.target_names)
plt.show()
```

```python
#1. Univariate
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#load tips dataset
df = sns.load_dataset('tips')

#1. Univariate Visualization

#histogram
plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'],bins=20,kde=True)
plt.title("Histogram of Total Bill")
plt.show()


#Boxplot
plt.figure(figsize=(6,4))
sns.boxplot(df['total_bill'])
plt.title("Boxplot of Total Bill")
plt.show()


#Pie chart
smoker_counts = df['smoker'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(smoker_counts,labels=smoker_counts.index, autopct='%1.1f%%')
plt.title("Smoker Distribution")
plt.show()



```

```python
#2. Bivariate Visulaizations

#Bivariate visualizations
#Scatter Plot

plt.figure(figsize=(6,4))
sns.scatterplot(x=df['total_bill'], y=df['tip'], hue=df['smoker'], palette='Set1')
plt.title('Total Bill vs Tip')
plt.show()

#Line chart
#Line Chart
plt.figure(figsize=(6,4))
plt.plot(df['total_bill'])
plt.title('Trend of Total Bill')
plt.xlabel('Sample Index')
plt.ylabel('Total Bill')
plt.show()

#bar chart
plt.figure(figsize=(6,4))
sns.barplot(x=df['day'], y=df['tip'])
plt.title("Bar chart: mean tip by day")
plt.show()
```

```python
#3. Multivariate Visualizations
#heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
plt.title('Heatmap of Tips Feature Correlations')
plt.show()

#bubble chart
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['total_bill'],y=df['tip'], s=df['size']*20, c=df['size'], alpha=0.5)
plt.title('Bubble Chart: Total Bill vs Tip with Party Size as Size')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()

#pair plot(relationships among all features)
sns.pairplot(df.select_dtypes(include='number'))
plt.suptitle('Pair Plot of Tips Numeric Features', y=1.02)
plt.show()
```

```python
#4.Advanced Techniques
#violin plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['day'], y=df['total_bill'])
plt.title('Violin Plot: Total Bill by Day')
plt.show()
```
